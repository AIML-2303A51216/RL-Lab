```python
gamma = 0.9

def state_value_function():
    rewards = [-1, -1, -1, 10]

    value = 0

    for i in range(len(rewards)):
        value += (gamma ** i) * rewards[i]

    return value


def action_value_function():
    rewards = [-1, -1, -1, -1, 10]

    value = 0

    for i in range(len(rewards)):
        value += (gamma ** i) * rewards[i]

    return value


print("State Value V(S0) =", state_value_function())
print("Action Value Q(S0, Up) =", action_value_function())
```
