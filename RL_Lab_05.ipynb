```python
# ============================================================
# ASSIGNMENT: BUILDING A RANDOM AGENT FOR REINFORCEMENT LEARNING
# ============================================================

# Install Gymnasium
!pip install -q "gymnasium[classic-control]"

# Import libraries
import gymnasium as gym


# ============================================================
# RANDOM AGENT
# ============================================================

def random_agent(environment_name):

    # 1. Create the selected environment
    env = gym.make(environment_name)

    # 2. Reset the environment
    state, info = env.reset()

    total_reward = 0
    step_count = 0

    print("\n" + "=" * 50)
    print("Selected Environment:", environment_name)
    print("=" * 50)

    # 3. Observe the initial state
    print("Initial State:", state)

    # 4. Identify available actions
    print("Available Actions:", env.action_space)

    # Run the episode
    while True:

        # 5. Select an action randomly
        action = env.action_space.sample()

        # Display the action
        print("Step:", step_count + 1, "| Action:", action)

        # 6. Execute action
        next_state, reward, terminated, truncated, info = env.step(action)

        # 7. Receive reward
        print("Reward:", reward)

        # Add reward
        total_reward += reward

        # Increase step count
        step_count += 1

        # 8. Observe next state
        state = next_state

        # Check whether episode is finished
        if terminated or truncated:
            break

    # 9. Calculate average reward
    average_reward = total_reward / step_count

    print("\nEpisode Finished")
    print("Total Steps:", step_count)
    print("Total Reward:", total_reward)
    print("Average Reward:", average_reward)

    # Close environment
    env.close()

    return total_reward, average_reward


# ============================================================
# DYNAMIC INPUT
# ============================================================

print("==========================================")
print("     RANDOM AGENT - REINFORCEMENT LEARNING")
print("==========================================")

print("\nAvailable Environments:")
print("1. CartPole")
print("2. MountainCar")
print("3. FrozenLake")

choice = input("\nEnter your choice (1/2/3): ")

# Select environment based on user input
if choice == "1":
    environment = "CartPole-v1"

elif choice == "2":
    environment = "MountainCar-v0"

elif choice == "3":
    environment = "FrozenLake-v1"

else:
    print("Invalid choice! Please enter 1, 2, or 3.")
    environment = None


# ============================================================
# RUN RANDOM AGENT
# ============================================================

if environment is not None:

    total_reward, average_reward = random_agent(environment)

    # Final result
    print("\n" + "=" * 50)
    print("FINAL RESULT")
    print("=" * 50)
    print("Environment:", environment)
    print("Total Reward:", total_reward)
    print("Average Reward:", average_reward)
```
